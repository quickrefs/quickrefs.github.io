### By position: iloc

#### Selecting one row by position

**B3.01**

`df.iloc[0]`

The `.iloc[]` indexer allows us to select a row by its integer position. It is useful when the first or last row is needed, regardless of labels. Alternatively, `.loc[]` selects by label.

`iloc` selects by **integer position** (0 is the first row), whatever the row labels are. One row comes out as a Series.

```python
# display the example data
df_students
```

```python
df_students.iloc[0]      # first row
```

```python
df_students.iloc[-1]     # last row
```

*See also:* Selecting a range of rows by position (B3) · Previewing the first and last rows (B1)

#### Selecting a range of rows by position

**B3.02**

`df.iloc[1:3]`

A slice inside `.iloc[]` allows us to select several consecutive rows by position (end excluded). It is useful for the first n rows. Alternatively, `df.head(n)` does the same for the start.

The end position is **excluded**: this gives rows 1 and 2 (the 2nd and 3rd).

```python
# display the example data
df_students
```

```python
df_students.iloc[1:3]
```

```python
df_students.iloc[:2]     # the first two rows
```

*See also:* Selecting one row by position (B3) · Selecting a range of rows by label (B3)

#### Selecting one cell by position

**B3.03**

`df.iloc[0, 1]`

Two numbers in `.iloc[row, col]` allow us to pick out one cell by position. It is useful for a single known value. Alternatively, `.iat[row, col]` is a faster version for single cells.

`iloc[row, column]`: both are positions.

```python
# display the example data
df_students
```

```python
df_students.iloc[0, 1]           # row 0, column 1
```

```python
df_students.columns[1]           # the name of column 1: this is the 'age' column
```

*See also:* Selecting one cell by label (B3) · Selecting rows and columns by position lists (B3) · Selecting one row by position (B3)

#### Selecting rows and columns by position lists

**B3.04**

`df.iloc[[0, 2], [0, 2]]`

Lists inside `.iloc[]` allow us to pick several non-adjacent rows and columns by position. It is useful for custom sub-tables. Alternatively, `.loc[]` with lists of labels does the same by name.

Give a list of positions for rows, and another for columns.

```python
# display the example data
df_students
```

```python
df_students.iloc[[0, 2], [0, 2]]       # rows 0 and 2; columns 0 and 2
```

```python
df_students.iloc[1:3, 1:]              # slices work too
```

*See also:* Selecting one cell by position (B3)

### By label: loc

#### Selecting one row by label

**B3.05**

`df.loc['a1']`

The `.loc[]` indexer allows us to select a row by its label. It is useful when rows have meaningful labels. Alternatively, `.iloc[]` selects by position.

`loc` selects by **label**. With the default index (0, 1, 2…) the labels are numbers; here they are `a1`…`a4`.

```python
# display the example data
df_students
```

```python
df_students.loc["a1"]
```

```python
df_students.loc["a9"]      # a label that does not exist
```

The available labels can be listed to find the mistake:

```python
df_students.index.tolist()
```

*See also:* Setting a column as the index (B2)

#### Selecting a range of rows by label

**B3.06**

`df.loc['a1':'a3']`

A slice inside `.loc[]` allows us to select consecutive rows by label, with the end **included**. It is useful for date ranges and ordered IDs. Alternatively, `.iloc[]` slices by position with the end excluded.

Unlike `iloc`, the end label is **included**.

```python
# display the example data
df_students
```

```python
df_students.loc['a1':'a3']     # a1, a2 AND a3
```

```python
df_students.iloc[0:3]          # positions 0, 1, 2 -> same rows, but end excluded
```

*See also:* Selecting one row by position (B3) · Selecting one row by label (B3)

#### Selecting one cell by label

**B3.07**

`df.loc['a2', 'age']`

The `.loc[row, column]` form allows us to pick out one cell by row label and column name. It is useful for looking up a single value. Alternatively, `.at[row, col]` is a faster version.

`loc[row label, column name]`

```python
# display the example data
df_students
```

```python
df_students.loc["a2", "age"]
```

*See also:* Selecting one cell by position (B3) · Selecting rows and columns by label lists (B3) · Selecting one row by label (B3)

#### Selecting rows and columns by label lists

**B3.08**

`df.loc[['a1', 'a3'], ['age', 'height']]`

Lists inside `.loc[]` allow us to pick specific rows and columns by name. It is useful for building a custom sub-table. Alternatively, `df[[cols]]` picks columns only.

```python
# display the example data
df_students
```

```python
df_students.loc[["a1", "a3"], ["age", "height"]]
```

```python
df_students.loc["a2":"a4", "age":"height"]     # label slices for columns too (end included)
```

*See also:* Selecting rows and columns by position lists (B3) · Filtering rows and choosing columns (B3)

### Boolean masks

#### Building a mask

**B3.09**

`mask = df['col'] > value`

A comparison on a column allows us to build a mask, a True/False value for every row. It is useful as the first step of filtering. Alternatively, `df.query()` writes the condition as text.

A comparison on a column gives a **mask**: a Series of `True`/`False`, one per row.

```python
# display the example data
df_students
```

```python
mask = df_students["age"] > 21
mask
```

Other tests work the same way:

```python
(df_students['name'] == 'Bo').tolist()
```

```python
(df_students['height'] <= 165).tolist()
```

*See also:* Filtering with `query` (B3) · Filtering rows with a mask (B3) · Comparing values (A2) · Combining conditions with `&` and `|` (B3)

#### Filtering rows with a mask

**B3.10**

`df[mask]  /  df.loc[mask]`

Putting a mask inside `df[...]` allows us to keep only the rows where it is `True`. It is useful for selecting a subset of the rows. Alternatively, `df.loc[mask]` does the same.

Only the rows where the mask is `True` are kept. The original row labels are kept too.

```python
mask = df_students["age"] > 21
mask.tolist()
```

```python
df_students[mask]
```

```python
df_students[df_students["age"] > 21]        # the same, in one line
```

Watch out: filtering keeps the old row labels, so call `reset_index(drop=True)` to renumber the rows from 0.

*See also:* Selecting one row by label (B3) · Building a mask (B3) · Discarding the old index (B2)

#### Filtering rows and choosing columns

**B3.11**

`df.loc[mask, ['age', 'height']]`

The form `df.loc[mask, columns]` allows us to filter rows and select columns in one step. It is useful when only a few variables are needed for a subset. Alternatively, filter first, then select columns in a second step.

`loc[rows, columns]` filters rows and picks columns in one step.

```python
mask = df_students["age"] > 21
df_students.loc[mask, ["name", "height"]]
```

*See also:* Filtering rows with a mask (B3) · Selecting rows and columns by label lists (B3) · Filtering rows in one line (B3)

#### Filtering rows in one line

**B3.12**

`df[df['col'] == value]`

Writing the condition inside the brackets allows us to filter in one line. It is useful for quick, one-off subsets. Alternatively, store the mask in a variable first if it is used more than once.

```python
df_adelie = df_penguins[df_penguins["species"] == "Adelie"]
print(f"{len(df_penguins)} penguins in total, {len(df_adelie)} are Adelie")
```

```python
# display example data
df_adelie.head(3)
```

*See also:* Building a mask (B3) · Combining conditions with `&` and `|` (B3) · Counting each category (B5) · Filtering with `query` (B3)

#### Combining conditions with `&` and `|`

**B3.13**

`(cond1) & (cond2),  (cond1) | (cond2)`

The `&` and `|` operators allow us to combine several masks, requiring all or any to be true. They are useful for filtering on more than one column. Alternatively, `isin()` tests one column against several values.

Use `&` (and) and `|` (or) rather than `and` and `or`, and wrap **each** condition in brackets.

```python
df_adelie_heavy = df_penguins[(df_penguins["species"] == "Adelie") & (df_penguins["body_mass_g"] > 4000)]
df_dream_or_biscoe = df_penguins[(df_penguins["island"] == "Dream") | (df_penguins["island"] == "Biscoe")]
print(f"Adelie and > 4000 g : {len(df_adelie_heavy)}")
```

```python
print(f"Dream or Biscoe     : {len(df_dream_or_biscoe)}")
```

Watch out: without the brackets, `&` is applied before the comparisons and an error results:

```python
df_penguins[df_penguins["species"] == "Adelie" & df_penguins["body_mass_g"] > 4000]
```

*See also:* Building a mask (B3) · Combining conditions with `and`, `or`, `not` (A2) · Filtering with `~`, `isin()` and `between()` (B3)

#### Copying a filtered table

**B3.14**

`df[mask].copy()`

The `.copy()` method allows us to turn a filtered view into its own independent table. It is useful before adding columns, to avoid `SettingWithCopyWarning`. Alternatively, use `.loc` for assignments when the original is meant to change.

After filtering, pandas cannot tell whether a separate table or a view into the original is intended. Adding a column to it can trigger a `SettingWithCopyWarning` (pandas 2.x). `.copy()` states that an independent table is wanted.

```python
df_gentoo = df_penguins[df_penguins["species"] == "Gentoo"].copy()
df_gentoo["mass_kg"] = df_gentoo["body_mass_g"] / 1000     # safe: gentoo is a real copy
df_gentoo[["species", "body_mass_g", "mass_kg"]].head(3)
```

*See also:* Copies and aliases of a list (A3) · Adding a constant column (B4) · Views and copies of an array slice (A4) · Checking a DataFrame after each step (B1)

#### Filtering with `~`, `isin()` and `between()`

**B3.15**

`~mask, s.isin(values), s.between(a, b)`

***(non-examinable)***

The `~`, `isin()` and `between()` tools allow us to negate a mask, match a list of values, or test a range. They are useful to shorten longer conditions. Alternatively, write the equivalent with `==`, `|` and two comparisons.

`~` flips a mask; `isin` matches any value in a list; `between` is an inclusive range test.

```python
print(f'not Adelie     : {len(df_penguins[~(df_penguins["species"] == "Adelie")])}')
```

```python
print(f'Adelie or Gentoo: {len(df_penguins[df_penguins["species"].isin(["Adelie", "Gentoo"])])}')
```

```python
print(f'3000-3500 g     : {len(df_penguins[df_penguins["body_mass_g"].between(3000, 3500)])}')
```

*See also:* Combining conditions with `&` and `|` (B3) · Building a mask (B3) · Filtering with `query` (B3)

`query` takes the condition as text. Masks remain the standard method in this book.

### Order of rows

#### Sorting by a column

**B3.17**

`df.sort_values(by='height')`

The `sort_values()` method allows us to order the rows by the values in a column. It is useful for rankings and for putting data in order. Alternatively, `.nlargest(n, 'col')` gives the top n directly.

```python
# display the example data
df_students
```

By default the smallest value comes first (ascending order).

```python
df_students.sort_values(by="height")
```

*See also:* Sorting in descending order (B3) · Sorting by several columns (B3) · Sorting by the index (B2)

#### Sorting in descending order

**B3.18**

`df.sort_values(by='height', ascending=False)`

The `ascending=False` option allows us to sort from largest to smallest. It is useful for top-10 style lists. Alternatively, `df.nlargest()` or `.head()` after sorting limits the output.

```python
df_students.sort_values(by="height", ascending=False)
```

Combined with `head`, this gives the largest few rows:

```python
df_students.sort_values(by="height", ascending=False).head(2)
```

*See also:* Sorting by a column (B3) · Sorting by several columns (B3) · Selecting rows and columns by label lists (B3)

#### Sorting by several columns

**B3.19**

`df.sort_values(by=['age', 'height'], ascending=[False, True])`

A list passed to `sort_values(by=[...])` allows us to sort by one column and break ties with another. It is useful for ordering groups and then values within them. Alternatively, sort twice with `kind="stable"`, starting with the less important column (the default sort does not keep the earlier order of ties).

Sort by the first column; ties are broken by the second. `ascending` can be a list, one value per column.

```python
# display the example data
df = pd.DataFrame({"group": ["a", "b", "a", "b"], "score": [3, 1, 2, 4]})
df
```

```python
# display example data
df
```

```python
df.sort_values(by=["group", "score"], ascending=[True, False])
```
