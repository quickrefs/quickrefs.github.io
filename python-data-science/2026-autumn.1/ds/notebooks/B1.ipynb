#### Creating a Series

**B1.02**

`pd.Series(values)`

A Series allows us to hold a single labelled column of values. It is useful when selecting a single column, because every column of a DataFrame is a Series. Alternatively, a NumPy array holds numbers without labels.

A **Series** is a single labelled column; a **DataFrame** is a table of Series that share the same index.

```python
s = pd.Series([165, 180, 172], name="height")
s
```

```python
type(s)
```

Custom labels are given with `index=`:

```python
pd.Series([165, 180, 172], index=["Ana", "Bo", "Cy"], name="height")
```

#### Creating a DataFrame from a dictionary

**B1.03**

`pd.DataFrame({'col': values})`

Passing a dictionary to `pd.DataFrame` allows us to build a table by hand, one column per key. It is useful for small examples and for calculation tables. Alternatively, load a table from a file with `pd.read_csv`.

Keys become column names. Each list becomes that column, so every list must be the same length.

```python
df_students = pd.DataFrame({
    "name":   ["Ana", "Bo", "Cy", "Di"],
    "age":    [21, 25, 22, 30],
    "height": [165, 180, 172, 158],
})
df_students
```

*See also:* Creating a dictionary (A3) · Creating a DataFrame with row labels (B1) · Creating a DataFrame from a list of rows (B1)

#### Creating a DataFrame with row labels

**B1.04**

`pd.DataFrame(data, index=labels)`

The `index=` argument allows us to label rows with our own names instead of 0, 1, 2…. It is useful when rows have meaningful identifiers like IDs or dates. Alternatively, create the table first and call `set_index()` on a column.

By default rows are labelled 0, 1, 2…. Supply `index=` to use custom labels, which can then be selected with `loc`.

```python
df_students = pd.DataFrame({
    "name":   ["Ana", "Bo", "Cy", "Di"],
    "age":    [21, 25, 22, 30],
    "height": [165, 180, 172, 158],
}, index=["a1", "a2", "a3", "a4"])
df_students
```

```python
df_students.index.tolist()
```

```python
df_students.loc["a3"]
```

#### Creating a DataFrame from a list of rows

**B1.05**

`pd.DataFrame(rows, columns=names)`

Passing a list of rows plus `columns=` allows us to build a DataFrame row by row. It is useful when the data naturally arrive as records. Alternatively, a dictionary of columns is often tidier.

Give one list per row, plus the column names.

```python
rows = [["Ana", 21, 165],
        ["Bo", 25, 180],
        ["Cy", 22, 172]]
pd.DataFrame(rows, columns=["name", "age", "height"])
```

#### Saving and reloading with `pickle`

**B1.18**

`df.to_pickle(f), pd.read_pickle(f)`

The `.to_pickle()` and `pd.read_pickle()` pair allows us to save a DataFrame exactly as it is, including types. It is useful for intermediate results that include dates or categories. Alternatively, CSV is more portable but loses type information.

Pickle keeps the data types and index exactly as they are, which CSV cannot do (for example dates and categories).

```python
# display the example data
df = pd.DataFrame({"when": pd.to_datetime(["2024-01-05", "2024-02-10"]), "v": [1, 2]})
df
```

```python
df.dtypes                                  # before saving: 'when' is a datetime
```

```python
df.to_pickle("df_demo.pkl")
```

```python
df_back = pd.read_pickle("df_demo.pkl")
df_back.dtypes                             # after reloading: the types are unchanged
```

A CSV file would return `when` as text:

```python
df.to_csv("df_demo.csv", index=False)
```

```python
pd.read_csv("df_demo.csv").dtypes
```

#### Checking a mask before using it

**B1.20**

`mask.sum(), mask.mean()`

***(non-examinable)***

Counting the True values of a mask allows us to check that a condition selects the rows that were intended. It is useful because a misspelt value selects no rows and gives an empty table without any error. Alternatively, `df[mask].shape` shows the size of the selection directly.

The text in a comparison must match exactly, including capital letters. Here no row matches:

```python
mask_adelie = df_penguins["species"] == "adelie"
mask_adelie.sum()                  # 0 rows selected
```

The exact values that are present can be listed:

```python
df_penguins["species"].unique()
```

With the correct spelling, the number and proportion of rows selected are:

```python
mask_adelie = df_penguins["species"] == "Adelie"
mask_adelie.sum()
```

```python
round(mask_adelie.mean(), 3)
```

#### Checking a new column against a hand calculation

**B1.21**

`df.loc[row, cols]`

***(non-examinable)***

Calculating one value by hand allows us to check that a whole column was calculated as intended. It is useful after any formula applied to a column, because a wrong unit or column gives plausible but wrong numbers. Alternatively, compare summary statistics (such as the mean) before and after the calculation.

```python
df_check = df_penguins.copy()
df_check["mass_kg"] = df_check["body_mass_g"] / 1000
df_check.loc[0, ["body_mass_g", "mass_kg"]]          # the first row
```

By hand, 3750 g / 1000 = 3.75 kg, which matches. A second check compares the means:

```python
round(df_check["body_mass_g"].mean() / 1000, 4)
```

```python
round(df_check["mass_kg"].mean(), 4)
```

#### Checking data with `assert`

**B1.22**

`assert df['c'].min() > 0`

***(non-examinable)***

An `assert` statement on a DataFrame allows us to state what must be true of the data, such as no negative values or no missing values. It is useful for stopping the analysis when a loaded file is not as expected. Alternatively, `df.describe()` and `df.isna().sum()` show the same information for inspection.

This check passes, so nothing is displayed:

```python
assert df_penguins["body_mass_g"].min() > 0, "body masses must be positive"
df_penguins["body_mass_g"].min()
```

This check fails because the column has missing values:

```python
assert df_penguins["bill_length_mm"].notna().all(), "bill_length_mm has missing values"
```
